```python
import numpy as np
import sympy as sp

a = sp.symbols('a')

#plate
c_plate = sp.Matrix([0,2*a])
w = 4*a
h = a/10
a_plate = w*h
I_plate = sp.Matrix([w*h**3/12, w**3 * h /12])

#z1
c_z1 = sp.Matrix([-a/2,0])
w = a
h = a/10
a_z1 = w*h
I_z1 = sp.Matrix([w*h**3/12, w**3 * h /12])

#z2
c_z2 = sp.Matrix([0,a])
w = a/10
h = 2*a
a_z2 = w*h
I_z2 = sp.Matrix([w*h**3/12, w**3 * h /12])

#z3
c_z3 = sp.Matrix([a/2, 2*a])
w = a
h = a/10
a_z3 = w*h
I_z3 = sp.Matrix([w*h**3/12, w**3 * h /12])

#tot
a_tot = a_plate + a_z1 + a_z2 +a_z3
c_tot = (
    c_plate * a_plate + 
    c_z1 * a_z1 + 
    c_z2 * a_z2 + 
    c_z3 * a_z3
) /a_tot

d_plate = c_plate - c_tot
d_z1 = c_z1 - c_tot
d_z2 = c_z2 - c_tot
d_z3 = c_z3 - c_tot

I_tot_dia =(
    I_plate + sp.Matrix([a_plate * d_plate[1]**2, a_plate * d_plate[0]**2]) +
    I_z1 + sp.Matrix([a_z1 * d_z1[1]**2, a_z1 * d_z1[0]**2]) +
    I_z2 + sp.Matrix([a_z2 * d_z2[1]**2, a_z2 * d_z2[0]**2]) +
    I_z3 + sp.Matrix([a_z3 * d_z3[1]**2, a_z3 * d_z3[0]**2])
)

I_yz_tot = (
    a_plate * d_plate[0] * d_plate[1] +
    a_z1 * d_z1[0] * d_z1[1] +
    a_z2 * d_z2[0] * d_z2[1] +
    a_z3 * d_z3[0] * d_z3[1]
)

I = sp.Matrix([[I_tot_dia[0],I_yz_tot],[I_yz_tot,I_tot_dia[1]]])
I
```
